```python
layer_colors = {
    'ASC': "#E0E084",  
    'CA3GL': '#4B99A6',   
    'CLAGL': '#9E9F3E',   
    'CTGL': '#2C4029',   
    'D1MSN': '#89E5F2',     
    'D2MSN': '#51A9B9',  
    'ITL23GL': '#92EA80',   
    'ITL4GL': '#519E3E',  
    'ITL5GL': "#3F6E35",   
    'ITL6GL': "#396131",      
    'L6bGL': '#33502D',  
    'LAMGA': '#7F542B',   
    'LSXGA': '#F0B4B4',   
    'MSGA': '#C53A32',    
    'NPGL': '#F9DDC1',  
    'OGC': "#DDF0F1",  
    'OPC': '#3F7FBF',  
    'PER': '#F39C12', 
    'PTGL': '#CCA79E', 
    'PVGA': '#CD6155',  
    'RGL': '#8E44AD',  
    'SSTGA': '#D57DBE', 
    'VIPGA': '#E67E22', 
    'VLMC': '#3B74AE',  
}
```

```python
import numpy as np
def rotate_xy(adata, angle_deg, obsm_key="spatial"):

    coords = adata.obsm[obsm_key]
    assert coords.shape[1] >= 2, "Coordinates must have at least x and y dimensions."

    angle_rad = np.deg2rad(angle_deg)

    R = np.array([
        [np.cos(angle_rad), -np.sin(angle_rad)],
        [np.sin(angle_rad),  np.cos(angle_rad)]
    ])

    xy = coords[:, :2]
    xy_rotated = xy @ R.T  

    coords_rotated = np.hstack([xy_rotated, coords[:, 2:]])
    adata.obsm[obsm_key] = coords_rotated

    return adata

adata_all.obsm['spatial'][:,-1] *= -1
adata_all.obsm['spatial'][:,-2] *= -1
adata_all = rotate_xy(adata_all, angle_deg=-90)
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
import matplotlib.cm as cm
import matplotlib.colors as mcolors

coords = adata_all.obsm['spatial']  
labels = adata_all.obs['annotation'].astype(str)
unique_labels = labels.unique()

unique_labels = labels.unique()
cmap = cm.get_cmap('tab20', len(unique_labels))  
label_to_color = {label: mcolors.rgb2hex(cmap(i)) for i, label in enumerate(unique_labels)}
cell_colors = labels.map(label_to_color)
label_to_color = {label: layer_colors.get(label, '#D3D3D3') for label in unique_labels}

cell_colors = labels.map(label_to_color)

fig = plt.figure(figsize=(16, 12))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    coords[:, 2], coords[:, 1], 1*coords[:, 0],  
    c=cell_colors,
    s=10,     
    alpha=0.8  
)

ax.set_title('')
ax.set_xlabel('')
ax.set_ylabel('')
ax.set_zlabel('')

ax.xaxis.pane.fill = False
ax.yaxis.pane.fill = False
ax.zaxis.pane.fill = False
ax.grid(False)

ax.set_xticks([])
ax.set_yticks([])
ax.set_zticks([])

ax.set_facecolor('none')
fig.patch.set_facecolor('none')

for axis in ['x', 'y', 'z']:
    getattr(ax, f'{axis}axis').set_pane_color('none')

ax.view_init(elev=190, azim=140)

plt.tight_layout()
plt.show()
```
